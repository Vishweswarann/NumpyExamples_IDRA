# Numpy Excercises

### Creation of one dimensional numpy array

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4 ,5])
print("Array 1: ", arr1)

arr2 = np.arange(1, 10, 2)
print("Array 2: ", arr2)

arr3 = np.linspace(1, 50, 5)
print("Array 3: ", arr3)

arr4 = np.random.rand(5)
print("Array 4: ", arr4)

arr5 = np.random.randint(1, 10, 3)
print("Array 5: ", arr5)

arr6 = np.zeros(5)
print("Array 6: ", arr6)

arr7 = np.ones(8)
print("Array 7: ", arr7)

arr8 = np.full(8, 9)
print("Array 8: ", arr8)
```

### Creating multi dimensional arrays

```python
arr1 = np.array([[1, 2], [3, 4]])
print("Two dimensional array: \n", arr1)

arr2 = np.arange(0, 9)
print("Using reshape: \n", arr2.reshape(3, 3))

arr3 = np.ones((2,2))
print("Ones: \n", arr3)

arr4 = np.random.randint(1, 100, size=(3, 3))
print("randint: \n", arr4)

arr5 = np.full((2, 2), 5)
print(arr5)

arr6 = np.random.randint(1, 100, size=(3, 3, 3))
print(arr6)
```

### Properties of arrays

```python
arr1 = np.arange(1, 10).reshape(3, 3)

print("Dimensions: ", arr1.ndim)
print("Shape: ", arr1.shape)
print("size: ", arr1.size)
print("Data type: ", arr1.dtype)

arr1 = arr1.astype(np.float32)
print("Data type: ", arr1.dtype)

```

### Indexing and slicing

```python
arr1 = np.arange(10)
arr2 = np.arange(1, 17).reshape(4, 4)

print(arr1[5])
print(arr1[3:7])
print(arr1[::2])
print(arr1[::-1])

print(arr2[1])
print(arr2[:,3])
print(arr2[2,1])
print(arr2[:2,2:])
print(arr2[1:3,1:3])
print(arr2[::3,::3])
```

### Boolean Masking

```python
arr1 = np.array([-5, 2, 8, -1, 0, 7, -3, 10, 4, -8])

arr2 = np.random.randint(1, 100, size=(4, 4))

print("Less than 0: ", arr1[arr1 < 0])
print("Even numbers: ", arr1[arr1 % 2 == 0])
print("Greater than 0 and less than 6: ", arr1[(arr1 > 0) & (arr1 < 6)])
print("Exactly 0 or 10: ", arr1[(arr1 == 0) | (arr1 == 10)])


print("Greater than 80: ", arr2[arr2 > 80])
print("Replacing every number less than 30 with 0")
arr3 = arr2.copy()
arr3[arr3 < 30] = 0

print(arr3)

print("count of numbers greater than 50: ", (arr2 > 50).sum())

arr3 = np.arange(27).reshape(3, 3, 3)

print("number greater than 15: ", arr3[arr3 > 15])
```

### Axis on aggreagating functions

```python
arr1 = np.random.rand(5, 24)
# print(arr1)

print("Maximum of each row: ", arr1.max(axis=1))
print("Shape: ", arr1.max(axis=1).shape)

print("Average of each column: ", arr1.mean(axis = 0))
```

```python
arr = np.random.randint(2000, 15000, size=(12, 4 , 7))

print(arr.sum(axis = 2))
print(arr.sum(axis = 2).shape)
print(arr.mean(axis = 0).mean(axis = 0))
print(arr.mean(axis = 0).mean(axis = 0).shape)
```

### Concatenation and clipping

```python
arr1 = np.arange(9).reshape(3, 3)
arr2 = np.arange(9, 18).reshape(3, 3)

# print(np.concatenate(arr1, arr2))
print(np.concatenate((arr1, arr2), axis = 0))
print()
print(np.concatenate((arr1, arr2), axis = 1))


arr3 = np.arange(1, 20)
arr3 = np.clip(arr3 + 20, 10, 25)
print(arr3)
```

### Broadcasting

```python
arr1 = np.linspace(1, 100, 25)
arr2 = np.linspace(1, 100, 25)

print(arr1 + arr2)
print()
print(arr1 ** arr2)
print()
print(np.sqrt(arr1))
print()
print(np.sin(arr1))
```
